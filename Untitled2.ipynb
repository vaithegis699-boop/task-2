```python
import matplotlib.pyplot as plt

df["sex"].value_counts().plot(kind="bar")

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Students")
plt.show()
```

```python
df["studytime"].value_counts().sort_index().plot(kind="bar")

plt.title("Study Time Distribution")
plt.xlabel("Study Time")
plt.ylabel("Number of Students")
plt.show()
```

```python
df["G3"].value_counts().head(10).plot(kind="pie", autopct="%1.1f%%")

plt.title("Top 10 Final Grades")
plt.ylabel("")
plt.show()
```

```python
df["age"].value_counts().sort_index().plot(kind="bar")

plt.title("Student Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Students")
plt.show()
```

```python
df.groupby("sex")["absences"].mean().plot(kind="bar")

plt.title("Average Absences by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Absences")
plt.show()
```

```python
plt.scatter(df["G1"], df["G3"])

plt.title("First Period Grade vs Final Grade")
plt.xlabel("G1 Grade")
plt.ylabel("G3 Final Grade")
plt.show()
```

```python
print("1. Average Final Grade:")
print(df["G3"].mean())

print("\n2. Average Absences:")
print(df["absences"].mean())

print("\n3. Average Grade by Gender:")
print(df.groupby("sex")["G3"].mean())

print("\n4. Average Grade by Failures:")
print(df.groupby("failures")["G3"].mean())

print("\n5. Average Grade by Study Time:")
print(df.groupby("studytime")["G3"].mean())
```
